## Creating Function to Scrape Wikipedia

```python
def scrape_wikipedia(article):
    url = f"https://en.wikipedia.org/wiki/{article}"

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"}
    )

    soup = BeautifulSoup(response.text, "html.parser")

    paragraphs = soup.find_all("p")

    text = " ".join(
        p.get_text(strip=True)
        for p in paragraphs
    )

    return text
```

## collecting data 

```python
import pandas as pd
import time

dataset = []

for label, articles in categories.items():

    print(f"\nCollecting {label} articles...")

    for article in articles:

        try:
            text = scrape_wikipedia(article)

            dataset.append({
                "Text": text,
                "Label": label
            })

            print("Done:", article)

            time.sleep(1)

        except Exception as e:
            print("Error:", article, e)
```

## Part 2 — Clean the Text

## Applying these steps-
- Remove HTML tags
- Remove citations like [1], [23]
- Remove punctuation/special characters
- Convert text to lowercase
- Remove extra spaces

## 1.Load the dataset

```python
import pandas as pd

df = pd.read_csv("wikipedia.csv")
print(df.head())
```

```python
## sample Text
sample_text = df["Text"][0]

print(sample_text[:500])
```

## Remove HTML Tags

```python
import re
## function
def remove_html(text):
    return re.sub(r'<.*?>', '', str(text))
```

```python
df["Text"] = df["Text"].apply(remove_html) ## apply to dataset
```

```python
## eg.
example = ("<p>I am doing study right now <b>python</b></p>")

print("Before:",example)
print("After removing HTML Tag:",remove_html(example))
```

## Remove citations like [1], [23]

```python
## func
def remove_citations(text):
    return re.sub(r'\[\d+\]', '', text)
```

```python
## e.g.
example = "Artificial Intelligence is important [1] and useful [23][50]."

print("Before:",example)
print("After removing citations:",remove_citations(example))
```

## Remove punctuation/special characters

```python
## func
def remove_punctuation(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)
```

```python
## e.g.
example = "Wow!!! Are you sure this works @ 100% — because I just tested it & it returned: Success  (time: 12:30 PM, cost: $0.00)."

print("Before:",example)
print("After removing punctuations/ Special Characters:",remove_punctuation(example))
```

## Convert text to lowercase

```python
def convert_lowercase(text):
    return text.lower()
```

```python
example = "Artificial Intelligence and Machine Learning is very Important"

print("Before:",example)
print("After converting to lowercase:",convert_lowercase(example))
```

## Remove extra spaces

```python
def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()
```

```python
example = "artificial     intelligence      is powerful"

print("Before:",example)
print("After removing space:",remove_extra_spaces(example))
```

## Step 2: Remove Stopwords

## 1.Download & import stopwords

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
```

## 2.creating stopword list

```python
stop_words = set(stopwords.words('english'))
print(list(stop_words)[:20])
```

## 3.Removing stopwords from tokens

```python
df['tokens_no_stopwords'] = df['tokens'].apply(
    lambda tokens: [word for word in tokens
                    if word.lower() not in stop_words]
)
```

## Step 3: Stemming vs Lemmatization

## Apply both:
- Stemming (PorterStemmer)
- Lemmatization (WordNetLemmatizer)
## Create:
- tokens_stemmed
- tokens_lemmatized
## Compare results on 10 words
## Answer: Which method is better and why?

## 1.Importing libraries

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer, WordNetLemmatizer
```

## 2.Initialize stemmer and lemmatizer

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

## 3.create new columns

```python
df['tokens_stemmed'] = df['tokens_no_stopwords'].apply(
    lambda tokens: [stemmer.stem(word) for word in tokens]
)

df['tokens_lemmatized'] = df['tokens_no_stopwords'].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

## 4. Compare results on 10 words

```python
sample_words = df['tokens_no_stopwords'].iloc[0][:10]

comparison = []

for word in sample_words:
    comparison.append([
        word,
        stemmer.stem(word),
        lemmatizer.lemmatize(word)
    ])

for original, stemmed, lemmatized in comparison:
    print(f"{original:15} {stemmed:15} {lemmatized}")
```

## 5.view result 

```python
df[['tokens_no_stopwords',
    'tokens_stemmed',
    'tokens_lemmatized']].head()
```

## Which method is better and why?
- Lemmatization is generally better for NLP tasks because:
- Lemmatization converts words to their actual base form. e.g. running → run
- gives meaningful words
- Gives more accuracy.
- stemming : stemming is faster but sometimes it creates incomplete words.
- use stemming when speed is imp and use lemmatization for NLP tasks

## Part4 : Convert Text to Numbers (TF-IDF)

- Split data: 80% train / 20% test
- Use TfidfVectorizer
## Steps:
- Fit only on training data
- Transform both train and test
## Show:
- Shape of TF-IDF matrix
- Top 15 important words per category

## 1.train-test-split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['Label'],
    test_size=0.2,
    random_state=42,
    stratify=df['Label']
)
```

## 2.TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

## 3. shape of TF-IDF Matrix

```python
print("Train TF-IDF shape:", X_train_tfidf.shape)
print("Test TF-IDF shape:", X_test_tfidf.shape)
```

## 4.get word names

```python
feature_names = vectorizer.get_feature_names_out()
```

## 5. Top 15 imp words per category

```python
import numpy as np

y_train = np.array(y_train)
categories = np.unique(y_train)

for category in categories:
    idx = (y_train == category)   # now it's a numpy boolean array
    
    mean_tfidf = X_train_tfidf[idx].mean(axis=0).A1
    
    top_indices = mean_tfidf.argsort()[-15:][::-1]
    top_words = [feature_names[i] for i in top_indices]
    
    print("Top words for category:", category)
    print(top_words)
```
